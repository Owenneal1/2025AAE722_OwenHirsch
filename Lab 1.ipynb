```python
import pandas as pd
import numpy as np

# pandas: A powerful library for data manipulation and analysis.
# Core features: DataFrames (tabular data), easy data cleaning, grouping, merging, and time series support.
# Applications in graduate economics: Handling survey data, panel data, time series, and merging datasets.


# Create a simple DataFrame
data = {'Country': ['USA', 'Canada', 'Germany'], 'GDP': [21.43, 1.84, 3.86]}
df = pd.DataFrame(data)
print("Pandas DataFrame example:")
print(df)

# numpy: A fundamental library for numerical computing in Python.
# Core features: Fast array operations, linear algebra, random number generation, and mathematical functions.
# Applications in graduate economics: Simulations, econometric computations, matrix algebra, and optimization.


# Create a numpy array and perform a calculation
arr = np.array([1, 2, 3, 4])
print("\nNumpy array example:")
print("Original array:", arr)
print("Array squared:", arr ** 2)
```

```python
import matplotlib.pyplot as plt

# Scatter plot: age vs wage
plt.scatter(df['age'], df['wage'], label='Data')

# Best fit line
import numpy as np
coeffs = np.polyfit(df['age'], df['wage'], 1)
best_fit = np.poly1d(coeffs)
plt.plot(df['age'], best_fit(df['age']), color='red', label='Best Fit Line')

plt.xlabel('Age')
plt.ylabel('Income')
plt.title('Age vs Income')
plt.legend()
plt.show()
```
