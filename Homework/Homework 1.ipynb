```python

import pandas as pd

# Download the 'Advertising' dataset if not already present
url = "https://www.statlearning.com/s/Advertising.csv"
advertising = pd.read_csv(url, index_col=0)
advertising.head()

import statsmodels.api as sm

# Define the predictor variables and the response variable
X = advertising[['TV', 'radio', 'newspaper']]
y = advertising['sales']

# Add a constant term to the predictor variables
X = sm.add_constant(X)

# Fit the OLS regression model
model = sm.OLS(y, X).fit()

# Show the coefficients, standard errors, t-statistics, and p-values
results_df = pd.DataFrame({
	"coef": model.params,
	"std err": model.bse,
	"t": model.tvalues,
	"P>|t|": model.pvalues
})
results_df

```

```python
# 4 . a

import numpy as np


rng = np.random.default_rng(1)
x = rng.normal(size=100)
y = 2 * x + rng.normal(size=100)

# Perform a simple linear regression of y onto x, without an intercept

beta = np.sum(x * y) / np.sum(x ** 2)
import statsmodels.formula.api as smf

# Fit a linear regression model without intercept using ModelSpec (formula interface)
model_no_intercept = smf.ols(formula="y ~ x - 1", data={"x": x, "y": y}).fit()
print("Estimated coefficient (slope) using statsmodels (no intercept):", model_no_intercept.params['x'])

# Report the coefficient estimate, standard error, t-statistic, and p-value
print("Coefficient estimate (beta):", model_no_intercept.params['x'])
print("Standard error:", model_no_intercept.bse['x'])
print("t-statistic:", model_no_intercept.tvalues['x'])
print("p-value:", model_no_intercept.pvalues['x'])


#### The result is statistically significant, so therefore we can reject the null hypothesis that beta = 0. This is because the p-value is very low and the t-statistic is very high. The estimate of beta is also close to the true value of 2.
```

```python
#### 4. b Answering this question as if the intercept was included

# Fit a linear regression model with intercept using statsmodels
import statsmodels.formula.api as smf

model_with_intercept = smf.ols(formula="y ~ x", data={"x": x, "y": y}).fit()
print("Estimated coefficients (with intercept):")
print(model_with_intercept.params)
print("\nStandard errors:")
print(model_with_intercept.bse)
print("\nt-statistics:")
print(model_with_intercept.tvalues)
print("\np-values:")
print(model_with_intercept.pvalues)


### the intercept is not statistically significant - meaning that we cannot reject the null hypothesis that the intercept is 0. In fact the intercept should be right around zero for the data because the normal distribution will be around 0 as a mean. The slope is statistically significant and very close to the true value of 2.
# 4. c the relationship between the results of a and b is the slopes and the intercepts are very zero - even when we exclude the intercept because the intercept is going through the origin anyways.

```

```python
import numpy as np

n = len(x)
beta_hat = np.sum(x * y) / np.sum(x ** 2)
se_beta = np.sqrt(np.sum((y - x * beta_hat) ** 2) / ((n - 1) * np.sum(x ** 2)))
t_stat1 = beta_hat / se_beta

# Alternative formula from the image
numerator = np.sqrt(n - 1) * np.sum(x * y)
denominator = np.sqrt(np.sum(x ** 2) * np.sum(y ** 2) - (np.sum(x * y)) ** 2)
t_stat2 = numerator / denominator

print("t-statistic (standard formula):", t_stat1)
print("t-statistic (alternative formula):", t_stat2)

```


The t-statistic for the slope $\beta_1$ in a simple linear regression (with intercept) is:

\[
t = \frac{\hat{\beta}_1}{\text{SE}(\hat{\beta}_1)}
\]

where

\[
\hat{\beta}_1 = \frac{\sum_{i=1}^n (x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^n (x_i - \bar{x})^2}
\]

and

\[
\text{SE}(\hat{\beta}_1) = \sqrt{ \frac{1}{n-2} \cdot \frac{\sum_{i=1}^n (y_i - \hat{y}_i)^2}{\sum_{i=1}^n (x_i - \bar{x})^2} }
\]

This shows the algebraic form of the t-statistic for the regression slope.

## Algebraic Steps: t-statistic for Regression Without Intercept
Let's derive the alternative formula for the t-statistic for the slope in a regression of $y$ onto $x$ (no intercept):
1. **Start with the standard t-statistic formula:**
$$
t = \frac{\hat{\beta}}{\text{SE}(\hat{\beta})}
$$

2. **The slope estimate:**
$$
\hat{\beta} = \frac{\sum x_i y_i}{\sum x_i^2}
$$

3. **The standard error:**
$$
\text{SE}(\hat{\beta}) = \sqrt{\frac{\sum (y_i - \hat{\beta} x_i)^2}{(n-1) \sum x_i^2}}
$$

4. **Expand the numerator of the variance:**
$$
\sum (y_i - \hat{\beta} x_i)^2 = \sum y_i^2 - 2\hat{\beta}\sum x_i y_i + \hat{\beta}^2 \sum x_i^2
$$
But $\hat{\beta} = \frac{\sum x_i y_i}{\sum x_i^2}$, so:
$$
\sum (y_i - \hat{\beta} x_i)^2 = \sum y_i^2 - \frac{(\sum x_i y_i)^2}{\sum x_i^2}
$$

5. **Plug this into the SE formula:**
$$
\text{SE}(\hat{\beta}) = \sqrt{\frac{\sum y_i^2 - \frac{(\sum x_i y_i)^2}{\sum x_i^2}}{(n-1) \sum x_i^2}}
$$

6. **Combine into the t-statistic:**
$$
t = \frac{\hat{\beta}}{\text{SE}(\hat{\beta})} = \frac{\frac{\sum x_i y_i}{\sum x_i^2}}{\sqrt{\frac{\sum y_i^2 - \frac{(\sum x_i y_i)^2}{\sum x_i^2}}{(n-1) \sum x_i^2}}}
$$

7. **Simplify the expression:**
$$
t = \frac{\sqrt{n-1} \sum x_i y_i}{\sqrt{\sum x_i^2 \sum y_i^2 - (\sum x_i y_i)^2}}
$$

This is the alternative algebraic form for the t-statistic, making the algebra visible step by step.
