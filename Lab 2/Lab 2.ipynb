```python
import numpy as np

arr = np.arange(1, 7).reshape(2, 3)
print("Shape:", arr.shape)
print("Array + 2:\n", arr + 2)
```

```python
np.random.seed(42)
random_numbers = np.random.randn(5)
print("Random numbers:", random_numbers)
print("Mean:", random_numbers.mean())
print("Standard deviation:", random_numbers.std())
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3]
y = [4, 5, 6]

plt.scatter(x, y)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Scatter Plot Example")
plt.show()
```

```python
# Create a 4x4 array with values 0 to 15
array_4x4 = np.arange(16).reshape(4, 4)

# Print the second row (index 1)
print("Second row:", array_4x4[1])

# Print all elements greater than 10
print("Elements greater than 10:", array_4x4[array_4x4 > 10])
```
