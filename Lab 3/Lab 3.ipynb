```python
# Install ISLP if not already installed
%pip install ISLP

import pandas as pd
from ISLP import load_data

# Load the College dataset
college = load_data('College')

# Display the first 10 rows
display(college.head(10))

# List all column names
print("Column names:", list(college.columns))

# Brief description of the dataset and selected variables
print("""
The College dataset contains information about 777 colleges in the US. Each row represents a college, and columns provide various attributes such as tuition, enrollment, and faculty statistics.

Variable meanings:
- Outstate: Out-of-state tuition (in USD).
- Private: Whether the college is private ('Yes') or public ('No').
- Room.Board: Cost of room and board (in USD).
- PhD: Percentage of faculty with a PhD.
- Top10perc: Percentage of new students from the top 10% of their high school class.
""")
```

```python
from sklearn.linear_model import LinearRegression

# Prepare the data
X = college[['Top10perc']]
y = college['Outstate']

# Fit the linear regression model
model = LinearRegression()
model.fit(X, y)

# Display the coefficients
print("Intercept:", model.intercept_)
print("Coefficient for Top10perc:", model.coef_[0])

##### the relationship is that as one more percent fo the incoming class is from the top 10 percent of their high school, then tuition for the institution increases by 128 dollars.
```

```python
# Fit a multiple linear regression model predicting Outstate tuition using Top10perc, Room.Board, and PhD
X_multi = college[['Top10perc', 'Room.Board', 'PhD']]
y = college['Outstate']

model_multi = LinearRegression()
model_multi.fit(X_multi, y)

# Report the coefficients
print("Intercept:", model_multi.intercept_)
print("Coefficients:")
for name, coef in zip(X_multi.columns, model_multi.coef_):
	print(f"  {name}: {coef}")
	

### the intercept is negative meaning that this model predicts a school would pay students to go there if there was 0 percent PhD faculty, 0 percent of students from the top 10 percent of their class. For evey one percent increase of the faculty with a Phd the tution rasises by 5.6 dollars. If Room and board increases by 1 dollars out of state tuition is expectd to rise by 1.88 dollars showing that more expensive schools have more expensive lodging. The biggest is that for every 1 percentage increase in the top 10 percent of HS grads the tuition increases by 82 dollars. It seems like a college would preffer to have the top students from highschool over faculty with PhDs based on this one regression.
```

```python
# Add a quadratic term for Top10perc to the previous model
X_quad = X_multi.copy()
X_quad['Top10perc^2'] = X_quad['Top10perc'] ** 2

# Fit the new model
model_quad = LinearRegression()
model_quad.fit(X_quad, y)

# Report the coefficients
print("Intercept:", model_quad.intercept_)
print("Coefficients:")
for name, coef in zip(X_quad.columns, model_quad.coef_):
	print(f"  {name}: {coef}")

# Interpretation:
#### because the coefficient for the quadratic term is negative it shows that the there is diminishing returns to getting more top 10 percent students. However the first percent increases significantly increasing out of state tuition. 
```

```python
%pip install statsmodels

import statsmodels.api as sm
from statsmodels.stats.anova import anova_lm

# Fit the linear model (without quadratic term)
X_multi_sm = sm.add_constant(X_multi)
model_linear_sm = sm.OLS(y, X_multi_sm).fit()

# Fit the quadratic model (with quadratic term)
X_quad_sm = sm.add_constant(X_quad)
model_quad_sm = sm.OLS(y, X_quad_sm).fit()

# Compare models using ANOVA
anova_results = anova_lm(model_linear_sm, model_quad_sm)
display(anova_results)
print("F-statistic:", anova_results['F'][1])
print("p-value:", anova_results['Pr(>F)'][1])


### becasue the F value is large and the p value is small we reject the null hypothesis that the quadratic term is not useful for explaining changes in out of state tutition. So it is a useful term to keep in. 
```
